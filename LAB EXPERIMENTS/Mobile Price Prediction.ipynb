```python

import pandas as pd
from sklearn.ensemble import RandomForestClassifier

data = {
    'RAM': [2, 3, 4, 4, 6, 8, 8, 12],
    'Battery': [2000, 2500, 3000, 3500, 4000, 4500, 5000, 6000],
    'Storage': [16, 32, 32, 64, 64, 128, 128, 256],
    'Price_Range': [0, 0, 1, 1, 2, 2, 3, 3]
}

df = pd.DataFrame(data)
X = df[['RAM', 'Battery', 'Storage']]
y = df['Price_Range']

model = RandomForestClassifier(random_state=42)
model.fit(X, y)

mobile = pd.DataFrame([[8, 4500, 128]],
                      columns=['RAM', 'Battery', 'Storage'])

prediction = model.predict(mobile)[0]
print("Predicted Price Range:", prediction)
```
