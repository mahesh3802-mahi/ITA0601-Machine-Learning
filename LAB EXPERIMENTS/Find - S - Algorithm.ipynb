```python

import pandas as pd

# Load your dataset without treating the first row as a header
df = pd.read_csv('/content/ws.csv', header=None)

df.columns = ['Sky', 'AirTemp', 'Humidity',
              'Wind', 'Water', 'Forecast', 'EnjoySport']

print("Training Dataset:")
print(df)

# FIND-S Algorithm
h = None

for i, row in df.iterrows():
    if row['EnjoySport'].strip().lower() == 'yes':
        x = row.iloc[:-1].tolist()

        if h is None:
            h = x
        else:
            for j in range(len(h)):
                if h[j] != x[j]:
                    h[j] = '?'

        print("Hypothesis:", h)

print("\nMost Specific Hypothesis:")
print(h)
```
