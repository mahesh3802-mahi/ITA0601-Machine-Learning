```python

import pandas as pd

df = pd.read_csv('/content/ws.csv', header=None)
X = df.iloc[:, :-1].values
Y = df.iloc[:, -1].values

S = ['Ø'] * (len(X[0]))
G = [['?' for _ in S]]

for x, y in zip(X, Y):
    if str(y).strip().lower() == 'yes':
        for i in range(len(S)):
            if S[i] == 'Ø':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'
    else:
        for g in G:
            for i in range(len(x)):
                if g[i] == '?' and S[i] != '?' and S[i] != x[i]:
                    g[i] = S[i]

print("Specific Boundary (S):", S)
print("General Boundary (G):", G)
```
